```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
# Step i: Read dataset
df = pd.read_csv('salary_prediction.csv')

# Display first few rows
print("Dataset preview:\n", df.head())

# Define input (X) and output (y)
X = df[['years_of_experience']]   # independent variable
y = df['salary']                 # dependent variable

# Split dataset (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Step ii: Slope and Intercept
print("\nSlope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Step iii: Plot regression line with actual data
plt.scatter(X, y)  # actual data points
plt.plot(X, model.predict(X))  # regression line
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Simple Linear Regression")
plt.show()

# Step iv: Predict new values
new_experience = np.array([[5], [10]])  # example inputs
predicted_salary = model.predict(new_experience)

print("\nPredicted salaries:")
for exp, sal in zip(new_experience, predicted_salary):
    print(f"Experience: {exp[0]} years -> Predicted Salary: {sal}")
```

```python
# Step v: Model evaluation
y_pred = model.predict(X_test)

print("\nModel Evaluation Metrics:")
print("Mean Absolute Error (MAE):", metrics.mean_absolute_error(y_test, y_pred))
print("Mean Squared Error (MSE):", metrics.mean_squared_error(y_test, y_pred))
print("Root Mean Squared Error (RMSE):", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print("R2 Score:", metrics.r2_score(y_test, y_pred))
```
