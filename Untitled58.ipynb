```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt

# Load datasets
df1 = pd.read_csv("medical_records.csv")
df2 = pd.read_csv("patient_details.csv")
```

```python
# i. Merge datasets using inner join on patient_id
df = pd.merge(df1, df2, on="patient_id", how="inner")
print(df)
```

```python
# ii. Impute null values in name column
df['name'] = df['name'].fillna("unknown")
print(df['name'])
```

```python
# iii. Rectify negative values in age column
df['age'] = df['age'].apply(lambda x: abs(x) if x < 0 else x)
print(df['age'])
```

```python
# iv. Standardize gender column
df['gender'] = df['gender'].replace({
    'M': 'Male',
    'F': 'Female',
    'male': 'Male',
    'female': 'Female'
})
print(df['gender'])
```

```python
# v. Make city column case consistent (title case)
df['city'] = df['city'].str.title()
print(df['city'])
```

```python
# vi. Standardize admission_date format (YYYY-MM-DD)
df['admission_date'] = df['admission_date'].apply(
    lambda x: re.sub(r'(\d{2})/(\d{2})/(\d{4})', r'\3-\2-\1', str(x))
)
print(df['admission_date'])
```

```python
# vii. Remove string values from height_cm and convert to numeric
df['height_cm'] = df['height_cm'].astype(str).str.extract('(\d+)')
df['height_cm'] = pd.to_numeric(df['height_cm'], errors='coerce')
print(df['height_cm'])
# Impute missing height values with mean
df['height_cm'].fillna(df['height_cm'].mean(), inplace=True)
print(df['height_cm'])
```

```python
# viii. Replace "high" in bp column with "140/90"
df['bp'] = df['bp'].replace("high", "140/90")
print(df['bp'])
```

```python
# ix. Impute sugar_level with mean
df['sugar_level'].fillna(df['sugar_level'].mean(), inplace=True)
print(df['sugar_level'])
```

```python
# x. Group by gender and compute stats on visit_cost
grouped = df.groupby('gender')['visit_cost'].agg(['mean', 'min', 'max'])
print(grouped)
```

```python
# xi. Plot histogram on diagnosis column
df['diagnosis'].value_counts().plot(kind='bar')
plt.title("Diagnosis Distribution")
plt.xlabel("Diagnosis")
plt.ylabel("Count")
plt.show()
```

```python
#Final cleaned dataset preview
print("\nCleaned Dataset:\n", df.head())
```

```python
# -------- DFS FUNCTION --------
def dfs(graph, start, goal, visited=None, path=None):
    if visited is None:
        visited = set()
    if path is None:
        path = []

    visited.add(start)
    path.append(start)

    # Goal found
    if start == goal:
        return path

    # Traverse neighbors
    for neighbor in graph[start]:
        if neighbor not in visited:
            result = dfs(graph, neighbor, goal, visited, path.copy())
            if result:
                return result

    return None
```

```python
# -------- i. STATIC GRAPH --------
static_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E', 'F'],
    'C': ['A', 'D', 'F'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['B', 'D', 'F'],
    'F': ['C', 'D', 'E', 'B']
}

print("Static Graph:", static_graph)
# -------- iii. USER INPUT --------
start = input("Enter start node: ")
goal = input("Enter goal node: ")

# -------- iv. DFS ON STATIC GRAPH --------
path1 = dfs(static_graph, start, goal)

if path1:
    print("Path in static graph:", " -> ".join(path1))
else:
    print("No path found in static graph")
```

```python
# -------- ii. DYNAMIC GRAPH --------
dynamic_graph = {}

nodes = int(input("\nEnter number of nodes: "))

# Input node names
node_list = []
for i in range(nodes):
    node = input(f"Enter node {i+1}: ")
    dynamic_graph[node] = []
    node_list.append(node)

edges = int(input("Enter number of edges: "))
print("Enter edges(u v)")

# Input edges
for i in range(edges):
    u = input("Enter first node of edge: ")
    v = input("Enter second node of edge: ")

    # Undirected graph
    dynamic_graph[u].append(v)
    dynamic_graph[v].append(u)

print("Dynamic Graph:", dynamic_graph)

# -------- DFS ON DYNAMIC GRAPH --------
start2 = input("Enter start node for dynamic graph: ")
goal2 = input("Enter goal node for dynamic graph: ")

path2 = dfs(dynamic_graph, start2, goal2)

if path2:
    print("Path in dynamic graph:", " -> ".join(path2))
else:
    print("No path found in dynamic graph")
```
