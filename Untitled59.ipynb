```python
from collections import deque

# BFS function to get traversal and shortest path
def bfs(graph, start, goal):
    visited = set()
    queue = deque([[start]])   # store paths
    traversal = []

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node not in visited:
            visited.add(node)
            traversal.append(node)

            # Goal found → return path
            if node == goal:
                return traversal, path

            # Add neighbors
            for neighbor in graph[node]:
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)

    return traversal, None
```

```python
# i. STATIC GRAPH
# ----------------------------------
print("\n--- Static Graph BFS ---")

graph_static = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E', 'F'],
    'C': ['A', 'D', 'F'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['B', 'D', 'F'],
    'F': ['B', 'C', 'D', 'E']
}

start = input("Enter start node (Static Graph): ")
goal = input("Enter goal node (Static Graph): ")

traversal, path = bfs(graph_static, start, goal)

print("Traversal Order:", " -> ".join(traversal))

if path:
    print("Shortest Path:", " -> ".join(path))
else:
    print("No path found")

```

```python
# ii. DYNAMIC GRAPH
# ----------------------------------
print("\n--- Dynamic Graph BFS ---")

n = int(input("Enter number of nodes: "))
graph_dynamic = {}

# Input nodes
for i in range(n):
    node = input(f"Enter node {i+1}: ")
    graph_dynamic[node] = []

# Input edges
e = int(input("Enter number of edges: "))
for i in range(e):
    u = input("Enter first node of edge: ")
    v = input("Enter second node of edge: ")

    graph_dynamic[u].append(v)
    graph_dynamic[v].append(u)   # undirected graph

start = input("Enter start node (Dynamic Graph): ")
goal = input("Enter goal node (Dynamic Graph): ")

traversal, path = bfs(graph_dynamic, start, goal)

print("Traversal Order:", " -> ".join(traversal))

if path:
    print("Shortest Path:", " -> ".join(path))
else:
    print("No path found")
```
