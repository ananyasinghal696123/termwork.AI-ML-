```python
# DFS function to find path
def dfs(graph, start, goal, visited=None, path=None):
    if visited is None:
        visited = set()
    if path is None:
        path = []

    visited.add(start)
    path.append(start)

    # If goal is found
    if start == goal:
        return path

    # Traverse neighbors
    for neighbor in graph[start]:
        if neighbor not in visited:
            result = dfs(graph, neighbor, goal, visited, path.copy())
            if result:
                return result

    return None
```

```python
# i. STATIC GRAPH (Given)
# -------------------------------
print("\n--- Static Graph DFS ---")

graph_static = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E', 'F'],
    'C': ['A', 'D', 'F'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['B', 'D', 'F'],
    'F': ['C', 'D', 'E', 'B']
}

start = input("Enter start node (Static Graph): ")
goal = input("Enter goal node (Static Graph): ")

path = dfs(graph_static, start, goal)

if path:
    print("Path found:", " -> ".join(path))
else:
    print("No path found")
```

```python
# ii. DYNAMIC GRAPH (User Input)
# -------------------------------
print("\n--- Dynamic Graph DFS ---")

n = int(input("Enter number of nodes: "))
graph_dynamic = {}

# Initialize nodes
for i in range(n):
    node = input(f"Enter node {i+1}: ")
    graph_dynamic[node] = []

e = int(input("Enter number of edges: "))

# Add edges
for i in range(e):
    u = input("Enter first node of edge: ")
    v = input("Enter second node of edge: ")

    graph_dynamic[u].append(v)
    graph_dynamic[v].append(u)  # Undirected graph

start = input("Enter start node (Dynamic Graph): ")
goal = input("Enter goal node (Dynamic Graph): ")

path = dfs(graph_dynamic, start, goal)

if path:
    print("Path found:", " -> ".join(path))
else:
    print("No path found")
```
