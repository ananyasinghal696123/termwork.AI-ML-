```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# i. Load dataset
df = pd.read_csv("healthcare.csv")

print("\nOriginal Dataset:\n", df.head())

# Select only numeric columns for normalization
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

```

```python
# ii. Min-Max Scaling
minmax_scaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax[numeric_cols] = minmax_scaler.fit_transform(df[numeric_cols])

print("\nMin-Max Scaled Data:\n", df_minmax.head())

# iii. Standard Scaling
std_scaler = StandardScaler()
df_standard = df.copy()
df_standard[numeric_cols] = std_scaler.fit_transform(df[numeric_cols])

print("\nStandard Scaled Data:\n", df_standard.head())
```

```python
# iv. Plot graphs (Min-Max Scaling)
plt.figure()
df_minmax[numeric_cols].plot(kind='hist', bins=10)
plt.title("Min-Max Normalized Data (Histogram)")
plt.show()
```

```python
# v. Plot graphs (Standard Scaling)
plt.figure()
df_standard[numeric_cols].plot(kind='hist', bins=10)
plt.title("Standard Scaled Data (Histogram)")
plt.show()

# Additional comparison (Boxplot)
plt.figure()
df_minmax[numeric_cols].boxplot()
plt.title("Min-Max Scaling Boxplot")
plt.show()

plt.figure()
df_standard[numeric_cols].boxplot()
plt.title("Standard Scaling Boxplot")
plt.show()
```
