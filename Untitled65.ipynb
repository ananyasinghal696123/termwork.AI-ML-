```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# i. Create dataset
data = {
    "height": [170, 160, 180, 175, 165, 172, 168, 178, 182, 169],
    "weight": [65, 55, 75, 70, 60, 68, 62, 72, 80, 66],
    "gender": ["M", "F", "M", "F", None, "F", "M", None, "M", "F"]
}

df = pd.DataFrame(data)

# i. Compute BMI = weight(kg) / (height(m))^2
df["height_m"] = df["height"] / 100
df["BMI"] = df["weight"] / (df["height_m"] ** 2)

print("\nDataset with BMI:\n", df)
```

```python
# ii. Save dataset to CSV
df.to_csv("bmi_dataset.csv", index=False)

# iii. Load CSV into dataframe
df2 = pd.read_csv("bmi_dataset.csv")

# iv. Shape and size
print("\nShape:", df2.shape)
print("Size:", df2.size)
```

```python
# v. Count null values in gender
print("\nNull values in gender:", df2["gender"].isnull().sum())

# vi. Remove null values from gender column
df2 = df2.dropna(subset=["gender"])

print("\nAfter removing nulls:\n", df2)

# vii. Structural and statistical info
print("\nInfo:")
print(df2.info())

print("\nStatistical Summary:")
print(df2.describe())
```

```python
#viii. Histogram on gender column
plt.figure()
df2["gender"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# ix. Suitable plot (Height vs Weight)
plt.figure()
plt.scatter(df2["height"], df2["weight"])
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()

print("\nInference: Generally, weight increases with height (positive relationship).")
```

```python
# x. Aggregate functions
print("\nAggregate Functions:")
print("Height Mean:", df2["height"].mean())
print("Weight Mean:", df2["weight"].mean())
print("BMI Mean:", df2["BMI"].mean())

print("Height Max:", df2["height"].max())
print("Weight Min:", df2["weight"].min())
print("BMI Sum:", df2["BMI"].sum())
```

```python
# xi. Entries where BMI < 20
print("\nEntries with BMI < 20:\n", df2[df2["BMI"] < 20])
```
